```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report

from catboost import CatBoostClassifier

import joblib
```

```python
df = pd.read_excel("C:\\Data Science\\Debug\\ML\\data files\\KERALA R1 ALLOT.xlsx")
```

```python
df = df.loc[:, ~df.columns.str.contains("^Unnamed")]

df.dropna(axis=1, how="all", inplace=True)
```

```python
df.columns = [
    "Rank",
    "Candidate Category",
    "College Name",
    "Course Name"
]
```

```python
df.drop_duplicates(inplace=True)
```

```python
X = df[
    [
        "Rank",
        "Candidate Category",
        "Course Name"
    ]
]
```

```python
y = df["College Name"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
cat_features = [1, 2]
```

```python
model = CatBoostClassifier(
    iterations=1000,
    learning_rate=0.05,
    depth=8,
    loss_function="MultiClass",
    eval_metric="Accuracy",
    random_seed=42,
    verbose=100
)
```

```python
model.fit(
    X_train,
    y_train,
    cat_features=cat_features
)
```

```python
y_pred = model.predict(X_test)
```

```python
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy :", accuracy)
```

```python
print(classification_report(y_test, y_pred, zero_division=0))
```

```python
new_student = pd.DataFrame({
    "Rank": [rank],
    "Candidate Category": [category],
    "Course Name": [course]
})
```

```python
top_n = 5

# Predict probabilities
probabilities = model.predict_proba(new_student)

# Get Top 5 predictions
top_indices = probabilities[0].argsort()[-top_n:][::-1]

# Create Result DataFrame
result = pd.DataFrame({
    "Rank": range(1, top_n + 1),
    "College Name": [model.classes_[i] for i in top_indices],
    "Probability (%)": [
        round(probabilities[0][i] * 100, 2)
        for i in top_indices
    ]
})

print("\nTop 5 Predicted Colleges\n")
print(result)
```

```python
import pandas as pd

# -------------------------
# User Input
# -------------------------

rank = float(input("Enter Your Rank: "))
category = input("Enter Candidate Category: ").strip()
course = input("Enter Course Name: ").strip()

# -------------------------
# Filter Dataset
# -------------------------

filtered = df[
    (df["Candidate Category"] == category) &
    (df["Course Name"] == course)
].copy()

if filtered.empty:
    print("❌ No matching records found.")

else:

    # -------------------------
    # Calculate Gap
    # -------------------------

    filtered["Gap"] = filtered["Rank"] - rank

    # -------------------------
    # Find Best Match (Previous Year)
    # -------------------------

    closest_record = filtered.loc[filtered["Gap"].abs().idxmin()]

    best_match_college = closest_record["College Name"]
    best_match_rank = int(closest_record["Rank"])

    # -------------------------
    # Chance Function
    # -------------------------

    def chance(gap):
        if gap > 300:
            return "🟢 High Chance"
        elif gap >= -300:
            return "🟡 Borderline"
        else:
            return "🔴 Tough Chance"

    filtered["Chance"] = filtered["Gap"].apply(chance)

    # ---------------------------------------------------
    # Keep only one record per college
    # (Highest cutoff rank for that college)
    # ---------------------------------------------------

    result = (
        filtered
        .sort_values("Rank", ascending=False)
        .drop_duplicates(subset="College Name", keep="first")
    )

    # -------------------------
    # Sort by Rank (Ascending)
    # -------------------------

    result = result.sort_values("Rank").reset_index(drop=True)

    # -------------------------
    # Add Serial Number
    # -------------------------

    result.insert(0, "S.No", range(1, len(result) + 1))

    # -------------------------
    # Rename Columns
    # -------------------------

    result.rename(
        columns={
            "Candidate Category": "Category"
        },
        inplace=True
    )

    # -------------------------
    # Final Column Order
    # -------------------------

    result = result[
        [
            "S.No",
            "Course Name",
            "Category",
            "Rank",
            "College Name",
            "Chance"
        ]
    ]

    # -------------------------
    # Display Settings
    # -------------------------

    pd.set_option("display.max_columns", None)
    pd.set_option("display.max_colwidth", None)
    pd.set_option("display.width", 1500)

    # -------------------------
    # Prediction Summary
    # -------------------------

    print("\n" + "="*85)
    print("                    COLLEGE PREDICTION SUMMARY")
    print("="*85)
    print(f"Your Rank                  : {int(rank)}")
    print(f"Course                     : {course}")
    print(f"Category                   : {category}")
    print(f"Best Match (Previous Year) : {best_match_college}")
    print(f"Previous Year Cutoff Rank  : {best_match_rank}")
    print("="*85)

    # -------------------------
    # Display Result
    # -------------------------

    print("\nRecommended Colleges:\n")

    display(
        result.style
        .hide(axis="index")
        .set_properties(
            subset=["Course Name", "Category", "College Name", "Chance"],
            **{"text-align": "left"}
        )
        .set_properties(
            subset=["S.No", "Rank"],
            **{"text-align": "center"}
        )
        .set_table_styles([
            {
                "selector": "th",
                "props": [
                    ("text-align", "center"),
                    ("font-weight", "bold")
                ]
            }
        ])
    )
```
